```python
import pandas as pd
import plotly.express as px

orders = pd.read_csv("data/olist_orders_dataset.csv")
items = pd.read_csv("data/olist_order_items_dataset.csv")
customers = pd.read_csv("data/olist_customers_dataset.csv")
products = pd.read_csv("data/olist_products_dataset.csv")
translation = pd.read_csv("data/product_category_name_translation.csv")

delivered = orders[orders["order_status"] == "delivered"]
df = pd.merge(delivered, items, on="order_id")
df = pd.merge(df, customers, on="customer_id")
df = pd.merge(df, products[["product_id", "product_category_name"]], on="product_id", how="left")
df = pd.merge(df, translation, on="product_category_name", how="left")
df["product_category_name_english"] = df["product_category_name_english"].fillna("unknown")

df["order_purchase_timestamp"] = pd.to_datetime(df["order_purchase_timestamp"])
df["Month"] = df["order_purchase_timestamp"].dt.to_period("M")

df["cohort_month"] = df.groupby("customer_unique_id")["order_purchase_timestamp"].transform("min").dt.to_period("M")
df["months_since"] = (df["Month"].dt.year - df["cohort_month"].dt.year) * 12 + (df["Month"].dt.month - df["cohort_month"].dt.month)

snapshot = df["order_purchase_timestamp"].max() + pd.Timedelta(days=1)
rfm = df.groupby("customer_unique_id").agg(
    last_purchase=("order_purchase_timestamp", "max"),
    frequency=("order_id", "nunique"),
    monetary=("price", "sum")
)
rfm["recency"] = (snapshot - rfm["last_purchase"]).dt.days

df.shape
```

```python
rfm["R"] = pd.qcut(rfm["recency"], 5, labels=[5, 4, 3, 2, 1])
rfm["M"] = pd.qcut(rfm["monetary"], 5, labels=[1, 2, 3, 4, 5])
rfm["F"] = pd.cut(rfm["frequency"], bins=[0, 1, 2, float("inf")], labels=[1, 2, 3])
rfm.shape
```

```python
rfm[["recency", "frequency", "monetary", "R", "F", "M"]].head(10)
```

```python
rfm["R"] = rfm["R"].astype(int)
rfm["F"] = rfm["F"].astype(int)
rfm["M"] = rfm["M"].astype(int)
rfm["RFM_score"] = rfm["R"] + rfm["F"] + rfm["M"]
rfm[["R", "F", "M", "RFM_score"]].head()
```

```python
def segment(score):
    if score >= 12:
        return "Champions"
    elif score >= 9:
        return "Loyal"
    elif score >= 6:
        return "Potential"
    elif score >= 4:
        return "At Risk"
    else:
        return "Lost"

rfm["segment"] = rfm["RFM_score"].apply(segment)
rfm["segment"].value_counts()
```
